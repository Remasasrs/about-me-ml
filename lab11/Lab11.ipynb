```python
scaler = StandardScaler()
X_scaled = scaler.fit_transform(df)
```

```python
inertia_values = []

for k in range(1, 11):
    kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)
    kmeans.fit(X_scaled)
    inertia_values.append(kmeans.inertia_)

inertia_values
```

```python
plt.figure(figsize=(8, 6))
plt.plot(range(1, 11), inertia_values, marker="o")
plt.xlabel("Number of Clusters (K)")
plt.ylabel("Inertia")
plt.title("Elbow Method")
plt.show()
```

```python
silhouette_scores = []

for k in range(2, 11):
    kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)
    labels = kmeans.fit_predict(X_scaled)
    score = silhouette_score(X_scaled, labels)
    silhouette_scores.append(score)

silhouette_scores
```

```python
plt.figure(figsize=(8, 6))
plt.plot(range(2, 11), silhouette_scores, marker="o")
plt.xlabel("Number of Clusters (K)")
plt.ylabel("Silhouette Score")
plt.title("Silhouette Scores")
plt.show()
```

```python
silhouette_table = pd.DataFrame({
    "K": range(2, 11),
    "Silhouette Score": silhouette_scores
})

silhouette_table
```

```python
final_k = 4

final_kmeans = KMeans(n_clusters=final_k, random_state=42, n_init=10)
cluster_labels = final_kmeans.fit_predict(X_scaled)
```

```python
df["Cluster"] = cluster_labels
```

```python
df.head()
```

```python
cluster_summary = df.groupby("Cluster").mean()
cluster_summary
```

```python
df["Cluster"].value_counts().sort_index()
```

```python
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X_scaled)

pca_df = pd.DataFrame(X_pca, columns=["PC1", "PC2"])
pca_df["Cluster"] = df["Cluster"]

plt.figure(figsize=(8, 6))
sns.scatterplot(data=pca_df, x="PC1", y="PC2", hue="Cluster", palette="Set2")
plt.title("Customer Segments using PCA")
plt.show()
```

## Final Questions

1. This is an unsupervised learning problem because the dataset does not contain a target label. We are not predicting a known answer; instead, we are grouping customers based on similar behavior.

2. We removed the `CUST_ID` column because it is only an identification number. It does not describe customer behavior and should not affect clustering.

3. The columns that had missing values were `CREDIT_LIMIT` and `MINIMUM_PAYMENTS`.

4. The missing values were handled using mean imputation. This means each missing value was replaced with the mean value of its column.

5. Scaling is important before applying K-Means because K-Means depends on distance. Features with large values, such as `BALANCE`, `PURCHASES`, and `CREDIT_LIMIT`, can dominate the clustering if the data is not scaled.

6. I chose K = 4. The elbow method shows that the decrease in inertia starts to slow down around 4 clusters. The silhouette score was also checked to compare different K values. K = 4 gives a reasonable balance between cluster quality and meaningful customer segmentation.

7. Based on the cluster summary table, the customer segments can be described as:
   - Cluster 0: Moderate purchase customers with regular credit card usage.
   - Cluster 1: High-value customers with high purchases, high payments, and high credit limits.
   - Cluster 2: Customers who rely more on cash advance and have high balances.
   - Cluster 3: Low-activity customers with lower purchases and lower usage.

8. The cluster that may represent high-value customers is the cluster with the highest values in `PURCHASES`, `PAYMENTS`, and `CREDIT_LIMIT`.

9. The cluster that may represent customers who rely more on cash advance is the cluster with the highest values in `CASH_ADVANCE`, `CASH_ADVANCE_FREQUENCY`, and `CASH_ADVANCE_TRX`.

10. A company can use these clusters to create better marketing strategies. For example, high-value customers can receive premium offers, low-activity customers can receive engagement campaigns, and cash-advance-heavy customers can receive financial guidance or special repayment offers.
